```python
# MINI PROJECT : Student Data Explorer

import pandas as pd

# Student Dataset
data = {
    "name":["Ajay","Keerthana","Manoj","Swetha","Hari","Deepika","Rohit","Anitha"],
    "gender":["Male","Female","Male","Female","Male","Female","Male","Female"],
    "department":["CS","IT","ECE","CS","IT","ECE","CS","IT"],
    "math_score":[88,92,76,85,95,80,83,90],
    "programming_score":[91,94,72,87,96,81,85,93],
    "attendance_percentage":[84,95,70,89,97,73,86,91]
}
```

```python
# Create DataFrame
df = pd.DataFrame(data)
```

```python

# Calculate total score
df["total_score"] = df["math_score"] + df["programming_score"]
```

```python
# 1. Project Title
print("="*40)
print("STUDENT PERFORMANCE EXPLORER")
print("="*40)
```

```python
# 2. Dataset Overview
print("\nDataset Overview")

print("Total students:",len(df))
print("Total columns:",len(df.columns))
print("Unique departments:",df["department"].nunique())
```

```python

# 3. Department Wise Analysis
print("\nDepartment Analysis")

print(df["department"].value_counts())
```

```python
# 4. Score Analysis
print("\nScore Analysis")

print("Highest math score:",df["math_score"].max())
print("Lowest math score:",df["math_score"].min())
print("Average math score:",df["math_score"].mean())
print("Average programming score:",df["programming_score"].mean())
print("Average attendance:",df["attendance_percentage"].mean())
```

```python
# 5. Gender Wise Analysis
print("\nGender Wise Analysis")

print(df.groupby("gender")["total_score"].mean())
```

```python
# 6. Top Performers
print("\nTop 5 Students")

top_students = df.sort_values( by="total_score", ascending=False)
print(top_students[["name","department","total_score"]].head(5))
```

```python
# 7. Attendance Analysis
print("\nAttendance Analysis")

low_attendance = df[
df["attendance_percentage"] < 75
]

if len(low_attendance) > 0:

    print(
    low_attendance[
    ["name",
     "department",
     "attendance_percentage"]
    ]
    )

else:

    print(
    "All students attendance above 75%"
    )
```
